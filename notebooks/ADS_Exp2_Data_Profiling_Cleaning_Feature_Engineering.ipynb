Calendar Feature 1: Availability Rate

```python
calendar_availability_rate = (
    calendar_fe
    .groupby("listing_id")["available_numeric"]
    .mean()
    .reset_index(name="calendar_availability_rate")
)
```

Calendar Feature 2: Available Days

```python
calendar_available_days = (
    calendar_fe
    .groupby("listing_id")["available_numeric"]
    .sum()
    .reset_index(name="calendar_available_days")
)
```

Calendar Feature 3: Average Minimum Nights

```python
calendar_avg_minimum_nights = (
    calendar_fe
    .groupby("listing_id")["minimum_nights"]
    .mean()
    .reset_index(name="calendar_avg_minimum_nights")
)
```

Calendar Feature 4: Average Maximum Nights

```python

calendar_avg_maximum_nights = (
    calendar_fe
    .groupby("listing_id")["maximum_nights"]
    .mean()
    .reset_index(name="calendar_avg_maximum_nights")
)
```

Calendar Feature 5: Observation Count

```python
calendar_observation_count = (
    calendar_fe
    .groupby("listing_id")
    .size()
    .reset_index(name="calendar_observation_count")
)
```

Combine all 5

```python
calendar_features = calendar_availability_rate.merge(
    calendar_available_days,
    on="listing_id",
    how="outer"
)

calendar_features = calendar_features.merge(
    calendar_avg_minimum_nights,
    on="listing_id",
    how="outer"
)

calendar_features = calendar_features.merge(
    calendar_avg_maximum_nights,
    on="listing_id",
    how="outer"
)

calendar_features = calendar_features.merge(
    calendar_observation_count,
    on="listing_id",
    how="outer"
)

print("Calendar feature dataset shape:")
print(calendar_features.shape)

display(calendar_features.head(10))
```

## Reviews

```python
# Create a copy
reviews_fe = reviews_clean.copy()

# Convert review date
reviews_fe["date"] = pd.to_datetime(
    reviews_fe["date"],
    errors="coerce"
)

print("Reviews shape:", reviews_fe.shape)
print("\nColumns:")
print(reviews_fe.columns.tolist())
```

Feature 1 — Review Count

```python
review_count = (
    reviews_fe
    .groupby("listing_id")
    .size()
    .reset_index(name="review_count")
)
```

Feature 2 — Review Period

```python
review_dates = (
    reviews_fe
    .groupby("listing_id")["date"]
    .agg(
        first_review_date="min",
        last_review_date="max"
    )
    .reset_index()
)

review_dates["review_period_days"] = (
    review_dates["last_review_date"]
    - review_dates["first_review_date"]
).dt.days

review_period = review_dates[
    ["listing_id", "review_period_days"]
]
```

Feature 3 — Days Since Last Review

```python
reference_date = reviews_fe["date"].max()

review_recency = review_dates[
    ["listing_id", "last_review_date"]
].copy()

review_recency["days_since_last_review"] = (
    reference_date -
    review_recency["last_review_date"]
).dt.days

review_recency = review_recency[
    ["listing_id", "days_since_last_review"]
]
```

Feature 4 — Reviews Per Month


```python
reviews_per_month = review_dates.copy()

reviews_per_month["review_period_months"] = (
    reviews_per_month["review_period_days"] / 30
)

reviews_per_month["reviews_per_month_est"] = (
    reviews_fe.groupby("listing_id").size()
    .div(
        reviews_per_month
        .set_index("listing_id")["review_period_months"]
    )
)

reviews_per_month = (
    reviews_per_month
    .reset_index(drop=True)
    [["listing_id", "reviews_per_month_est"]]
)
```

```python
reviews_per_month["reviews_per_month_est"] = (
    reviews_per_month["reviews_per_month_est"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)
```

Feature 5 — Recent Review Count

```python
one_year_ago = reference_date - pd.Timedelta(days=365)

recent_review_count = (
    reviews_fe[
        reviews_fe["date"] >= one_year_ago
    ]
    .groupby("listing_id")
    .size()
    .reset_index(name="recent_review_count")
)
```

```python
reviews_features = review_count.merge(
    review_period,
    on="listing_id",
    how="outer"
)

reviews_features = reviews_features.merge(
    review_recency,
    on="listing_id",
    how="outer"
)

reviews_features = reviews_features.merge(
    reviews_per_month,
    on="listing_id",
    how="outer"
)

reviews_features = reviews_features.merge(
    recent_review_count,
    on="listing_id",
    how="outer"
)

# Listings with no recent reviews should have 0
reviews_features["recent_review_count"] = (
    reviews_features["recent_review_count"]
    .fillna(0)
)

print("Reviews feature dataset shape:")
print(reviews_features.shape)

display(reviews_features.head(10))
```

```python
review_selected_features = [
    "review_count",
    "review_period_days",
    "days_since_last_review",
    "reviews_per_month_est",
    "recent_review_count"
]

print("Missing values:")
print("=" * 60)

display(
    reviews_features[
        review_selected_features
    ].isna().sum()
)
```

```python
reviews_features[review_selected_features] = (
    reviews_features[review_selected_features]
    .fillna(0)
)
```
