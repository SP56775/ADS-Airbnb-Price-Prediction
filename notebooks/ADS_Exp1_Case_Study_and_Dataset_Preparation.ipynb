Step 14 — Perform an initial missing-value assessment

```python
def missing_value_summary(df):

    missing_count = df.isnull().sum()

    missing_percentage = (
        missing_count / len(df)
    ) * 100

    summary = pd.DataFrame({
        "Missing_Count": missing_count,
        "Missing_Percentage": missing_percentage
    })

    summary = summary[
        summary["Missing_Count"] > 0
    ]

    return summary.sort_values(
        by="Missing_Percentage",
        ascending=False
    )
```

```python
listings_missing = missing_value_summary(listings_df)

listings_missing.head(20)
```

```python
calendar_missing = missing_value_summary(calendar_df)

calendar_missing
```

```python
reviews_missing = missing_value_summary(reviews_df)

reviews_missing
```

Step 17 — Create a schema summary automatically

```python
def create_schema_summary(df, dataset_name):

    return pd.DataFrame({
        "Dataset": dataset_name,
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
        "Non_Null_Count": df.notnull().sum().values,
        "Missing_Count": df.isnull().sum().values,
        "Unique_Values": df.nunique(dropna=True).values
    })
```

```python
listings_schema = create_schema_summary(
    listings_df,
    "Listings"
)

calendar_schema = create_schema_summary(
    calendar_df,
    "Calendar"
)

reviews_schema = create_schema_summary(
    reviews_df,
    "Reviews"
)
```

```python
complete_schema = pd.concat(
    [
        listings_schema,
        calendar_schema,
        reviews_schema
    ],
    ignore_index=True
)

complete_schema.head()
```

```python
complete_schema.to_csv(
    os.path.join(
        OUTPUT_TABLE_PATH,
        "exp1_complete_schema_summary.csv"
    ),
    index=False
)
```

Step 18 — Document the integration strategy

## Dataset Integration Strategy

The Listings dataset will act as the master dataset because it contains one record per Airbnb listing.

The Calendar dataset has multiple records per listing because availability and price information are recorded for individual dates. Therefore, Calendar data cannot be directly merged with Listings. It will first be aggregated using `listing_id` to create listing-level features.

The Reviews dataset also contains multiple records per listing. It will be aggregated using `listing_id` to generate review activity and, where appropriate, sentiment-based features.

After aggregation, both datasets will contain one row per `listing_id` and can be merged with the Listings dataset.

Planned integration:

Listings
    +
Aggregated Calendar Features
    +
Aggregated Review Features
    ↓
Final Integrated Airbnb Dataset
